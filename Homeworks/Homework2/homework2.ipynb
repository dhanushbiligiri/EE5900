```python
from ultralytics import YOLO
import os
```

```python
dataset_path = "data"

if os.path.exists(dataset_path):
    print("Dataset folder found")
else:
    print("Dataset folder NOT found")
```

```python
model = YOLO("yolov8n.pt")

print("Model loaded successfully")
```

```python
results = model.train(
    data="data/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    name="license_plate_detector"
)
```

```python
metrics = model.val()
print(metrics)
```

```python
model.predict(
    source="data/test/images",
    conf=0.25,
    save=True
)
```

```python
import glob
from IPython.display import Image, display

prediction_images = glob.glob("runs/detect/predict/*.jpg")

print("Number of predicted images:", len(prediction_images))

for img in prediction_images[:5]:  # show first 5 images
    display(Image(filename=img))
```

```python
results = model.predict("data/test/images", conf=0.25)

for r in results:
    print("Image:", r.path)
    print("Boxes:", r.boxes.xyxy)
    print("Confidence:", r.boxes.conf)
    print("Class:", r.boxes.cls)
    print()
```

```python
results = model.predict("data/test/images", conf=0.25)

for r in results[:5]:
    im_array = r.plot()  # draw boxes
    display(Image(data=im_array))
```
